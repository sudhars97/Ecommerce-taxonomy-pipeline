```python
from datasets import load_dataset
import pandas as pd
import duckdb
import os

print("Downloading ESCI dataset from Hugging Face...")
# 1. Load the public dataset using the Hugging Face library
dataset = load_dataset("tasksource/esci")

# Convert the 'train' split into a standard Pandas DataFrame
df = dataset['train'].to_pandas()

# 2. Create the data directory
os.makedirs('data', exist_ok=True)

# 3. Save the first 100 rows to a CSV for your GitHub repo
sample_df = df.head(100)
sample_df.to_csv('data/first_100_rows_esci.csv', index=False)
print("Saved first_100_rows_esci.csv to the 'data' folder.")

# 4. Connect to an in-memory DuckDB database
con = duckdb.connect(database=':memory:')
con.register('products', df)

# QUERY 1: Find how many products are missing a brand name
missing_brand_query = """
    SELECT count(*) as missing_brands
    FROM products
    WHERE product_brand IS NULL OR product_brand = ''
"""
print("\nMissing Brands:")
print(con.execute(missing_brand_query).fetchdf())

# QUERY 2: Find products with very short descriptions (less than 20 characters)
short_desc_query = """
    SELECT product_id, product_title, product_description
    FROM products
    WHERE length(product_description) < 20
    LIMIT 5
"""
print("\nProducts with severely short descriptions:")
print(con.execute(short_desc_query).fetchdf())

# QUERY 3: Check the distribution of text lengths in product titles
title_length_query = """
    SELECT
        MIN(length(product_title)) as min_title_len,
        MAX(length(product_title)) as max_title_len,
        AVG(length(product_title)) as avg_title_len
    FROM products
"""
print("\nProduct Title Length Stats:")
print(con.execute(title_length_query).fetchdf())
```
