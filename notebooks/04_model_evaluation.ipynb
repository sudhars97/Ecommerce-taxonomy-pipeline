```python
import pandas as pd
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, classification_report

print("Initializing Evaluation Framework...")

# 2. Mocking the Evaluation Data
# In the real world, a Data Scientist creates a "Ground Truth" set manually.
# 'ground_truth_category' is what a human expert labeled the item.
# 'llm_predicted_category' is what our prompt from Notebook 02 outputted.
eval_data = {
    'product_title': [
        "Nike Men's Air Max 270 Black",
        "Spartan Industrial Reclosable Zip Bags",
        "Ray-Ban Aviator Classic",
        "Mens Summer Casual Shorts",
        "Apple iPhone 15 Pro Max Case",
        "EunWow Pirate Anchor Necklace",
        "Heavy Duty Packing Tape 6-Rolls"
    ],
    'ground_truth_category': [
        "Footwear",
        "Packaging",
        "Accessories",
        "Apparel",
        "Accessories",
        "Jewelry",
        "Packaging"
    ],
    'llm_predicted_category': [
        "Footwear",        # Correct
        "Packaging",       # Correct
        "Accessories",     # Correct
        "Apparel",         # Correct
        "Electronics",     # INCORRECT (LLM thought phone case was electronics)
        "Jewelry",         # Correct
        "Packaging"        # Correct
    ]
}

df_eval = pd.DataFrame(eval_data)

# 3. Calculate Core ML Metrics
# We use 'macro' averaging to treat all categories equally, regardless of how many items are in them.
y_true = df_eval['ground_truth_category']
y_pred = df_eval['llm_predicted_category']

accuracy = accuracy_score(y_true, y_pred)
precision = precision_score(y_true, y_pred, average='macro', zero_division=0)
recall = recall_score(y_true, y_pred, average='macro', zero_division=0)
f1 = f1_score(y_true, y_pred, average='macro', zero_division=0)

print("\n--- LLM TAXONOMY MODEL PERFORMANCE ---")
print(f"Accuracy:  {accuracy:.2%}")
print(f"Precision: {precision:.2f}")
print(f"Recall:    {recall:.2f}")
print(f"F1 Score:  {f1:.2f}")
print("-" * 38)

# 4. Generate a detailed Classification Report
print("\nDetailed Classification Report:")
print(classification_report(y_true, y_pred, zero_division=0))

# 5. Identify the Errors for Iterative Refinement
# A good Data Scientist always investigates WHY the model failed.
print("\nError Analysis (Where did the LLM fail?):")
errors = df_eval[df_eval['ground_truth_category'] != df_eval['llm_predicted_category']]
display(errors)
```
